```python
import statistics as stats
def compute_central_tendency(numbers):
  mean=stats.mean(numbers)
  median=stats.median(numbers)
  try:
      mode=stats.mode(numbers)
  except stats.StatisticsError:
      mode="No unique mode found"
  return mean,median,mode
def compute_dispersion(numbers):
  variance=stats.variance(numbers)
  std_dev=stats.stdev(numbers)
  return variance,std_dev
def main():
  numbers=[13,14,15,18,19,20,25,38]
  mean,median,mode=compute_central_tendency(numbers)
  variance,std_dev=compute_dispersion(numbers)

  print(f"Central tendency measure")
  print(f"Mean: {mean}")
  print(f"Median: {median}")
  print(f"Mode: {mode}")
  print(f"Variance: {variance}")
  print(f"std_dev: {std_dev}")
if __name__ == "__main__":
  main()
```
