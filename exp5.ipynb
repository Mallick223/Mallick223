```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
data={
      'size':[2104,1600,2400,1416,3000,1985,1534,1427,1380,1494],
       'Bedrooms':[3,3,3,2,4,4,3,3,3,3],
        'Age':[15,10,20,5,30,5,15,20,15,5],
        'Price':[39990,32990,36900,23200,53990,29990,31490,19899,21200,24250],
                }
dataset=pd.DataFrame(data)
X=dataset[['size','Bedrooms','Age']]
y=dataset['Price']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
mse=mean_squared_error(y_test,y_pred)
print("Predicted Prices:",y_pred)
print("Mean Squared Error:",mse)
```
