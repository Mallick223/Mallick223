### Simple Linear Regression Example

This example demonstrates linear regression using a very basic dataset and calculates the slope, intercept, and a prediction for a new value.

```python
import numpy as np
from scipy.stats import linregress
import matplotlib.pyplot as plt
import seaborn as sns

# Define a simple dataset (using the 'df' from the main analysis for habit differentiation)
x_data = df['Exercise_Hours'].values  # Independent variable
y_data = df['Sleep_Hours'].values    # Dependent variable

print("Input Data (Exercise Hours vs. Sleep Hours):")
for i in range(len(x_data)):
    print(f"Exercise = {x_data[i]:.1f}, Sleep = {y_data[i]:.1f}")

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(x_data, y_data)

print(f"\nLinear Regression Results:")
print(f"  Slope (m): {slope:.2f}")
print(f"  Intercept (b): {intercept:.2f}")
print(f"  R-squared: {r_value**2:.2f} (Explains {r_value**2*100:.0f}% of variance)")

# Make a prediction for a new x value (example: predict sleep for 5 hours of exercise)
new_x = 5.0
predicted_y = slope * new_x + intercept

print(f"\nPrediction:")
print(f"  For {new_x:.1f} hours of exercise, Predicted sleep = {predicted_y:.2f} hours")

# Plotting the simple regression line, differentiating habits
plt.figure(figsize=(8, 6))
sns.scatterplot(x=x_data, y=y_data, hue=df['True_Habit'],
                palette={'Good': '#00c853', 'Bad': '#ff1744'}, s=100, alpha=0.9)

# Plot the regression line
plt.plot(x_data, slope * x_data + intercept, color='blue', linewidth=3, label='Regression Line')

plt.title('Simple Linear Regression: Sleep Hours vs. Exercise Hours (by Habit)', fontsize=14, fontweight='bold')
plt.xlabel('Daily Exercise Hours', fontsize=12)
plt.ylabel('Daily Sleep Hours', fontsize=12)
plt.legend(title='Habit Type')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

print("\nAs shown in the plot, the single regression line attempts to find a general trend between exercise and sleep, but clearly separates the 'Good' and 'Bad' habit clusters, indicating distinct groups rather than a single linear relationship for the entire dataset.")
```
